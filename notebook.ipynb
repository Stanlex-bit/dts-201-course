# Class 1

```python
import statistics

ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# a. Number of people
number_of_people = len(ages)
print("Number of people:", number_of_people)

# b. Total of all ages
total_age = sum(ages)
print("Total of all ages:", total_age)

# c. Mean age
mean_age = total_age / number_of_people
print("Mean age:", round(mean_age, 1))

# d. Youngest and oldest ages
youngest = min(ages)
oldest = max(ages)
print("Youngest age:", youngest)
print("Oldest age:", oldest)

# e. Median
median_age = statistics.median(ages)
print("Median age:", median_age)

# Compare median and mean
print("Is the median different from the mean?", median_age != mean_age)
```

# Class 3

```python
import pandas as pd

data = pd.read_excel("50_startups_messy.xlsx")

print(data.shape)
print(data.isnull().sum())
print(data.duplicated().sum())
```

```python
# Drop duplicate rows
data = data.drop_duplicates()

# Standardise the State column
data["State"] = data["State"].str.strip().str.title()

# Convert Marketing Spend to numeric
data["Marketing Spend"] = pd.to_numeric(data["Marketing Spend"], errors="coerce")

print(data)
```

```python
simple_sample = data.sample(n=20, random_state=7)

print(simple_sample)
```

```python
stratified_sample = data.groupby("State", group_keys=False).sample(
    frac=0.20,
    random_state=7
)

print(stratified_sample)
```

```python
print("Population means:")
print(data[["R&D Spend", "Marketing Spend"]].mean())

print("\nSimple random sample means:")
print(simple_sample[["R&D Spend", "Marketing Spend"]].mean())

print("\nStratified sample means:")
print(stratified_sample[["R&D Spend", "Marketing Spend"]].mean())
```

the stratified sample is closer because its means have smaller differences from the population means
